# Exploración de `item`

## Importación de librerías y datos

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
item = pd.read_csv("../../data/processed/item.csv")
```

```python
print(f"Registros: {item.shape[0]}\nColumnas: {item.shape[1]}\n")
print(f"Columnas: {item.columns}\n")
```

## Estadísticas y visualizaciones

### Cantidad de películas por año de lanzamiento

```python
item['release_date'] = pd.to_datetime(item['release_date'], format='%Y-%m-%d', errors='coerce')
movies_per_year = item['release_date'].dt.year.value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.plot(movies_per_year.index, movies_per_year.values, marker="o", color="#4C72B0")
plt.xlabel("Año de lanzamiento")
plt.ylabel("Cantidad de películas")
plt.title("Cantidad de películas por año de lanzamiento")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Conteo y proporción de películas por género

```python
genre_columns = ["unknown", "Action", "Adventure", "Animation", "Children's", "Comedy",
                  "Crime", "Documentary", "Drama", "Fantasy", "Film-Noir", "Horror",
                  "Musical", "Mystery", "Romance", "Sci-Fi", "Thriller", "War", "Western"]

movies_per_genre = item[genre_columns].sum().sort_values(ascending=False)
genre_props = movies_per_genre / item.shape[0]

genre_summary = pd.DataFrame({
    "conteo": movies_per_genre,
    "proporcion": genre_props.round(4)
})
genre_summary
```
